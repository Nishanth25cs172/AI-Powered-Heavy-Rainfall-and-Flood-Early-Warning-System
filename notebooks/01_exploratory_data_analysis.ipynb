# AI-Powered Heavy Rainfall and Flood Early Warning System
## Notebook 01: Exploratory Data Analysis (EDA)

This notebook performs exploratory analysis of the project's official datasets:
1. **Flood Risk Dataset** (`data/raw/flood_risk.csv`)
2. **Daily Rainfall Dataset** (`data/raw/Daily Rainfall Data - India (2009-2024).csv`)
3. **Weather Prediction Dataset** (`data/raw/Indian Rainfall and Weather Prediction Dataset.xlsx`)

```python
import os
import pandas as pd
import numpy as np

BASE_DIR = os.path.dirname(os.getcwd()) if os.path.basename(os.getcwd()) == 'notebooks' else os.getcwd()
RAW_DATA_DIR = os.path.join(BASE_DIR, 'data', 'raw')
print(f"Raw Data Directory: {RAW_DATA_DIR}")
```

### 1. Flood Risk Dataset Analysis

```python
flood_path = os.path.join(RAW_DATA_DIR, 'flood_risk.csv')
df_flood = pd.read_csv(flood_path)
print(f"Flood Risk Dataset Shape: {df_flood.shape}")
df_flood.info()
df_flood.describe(include='all')
```

### 2. Historical Daily Rainfall Dataset Analysis

```python
rain_path = os.path.join(RAW_DATA_DIR, 'Daily Rainfall Data - India (2009-2024).csv')
df_rain = pd.read_csv(rain_path)
print(f"Daily Rainfall Dataset Shape: {df_rain.shape}")
df_rain['date'] = pd.to_datetime(df_rain['date'], errors='coerce')
print("Date Range:", df_rain['date'].min(), "to", df_rain['date'].max())
df_rain.describe()
```

### 3. Summary of Feature Bounds & Distributions

```python
print("=== Flood Dataset Numeric Summary ===")
for col in ['Rainfall (mm)', 'Water Level (m)', 'River Discharge (m³/s)', 'Elevation (m)', 'Latitude', 'Longitude']:
    if col in df_flood.columns:
        print(f"{col:25s} | Min: {df_flood[col].min():8.2f} | Max: {df_flood[col].max():8.2f} | Mean: {df_flood[col].mean():8.2f}")
```
