# Module A: Heavy Rainfall Weather Dataset — Exploratory Data Analysis & Data Quality Audit

This notebook performs exploratory analysis, date range verification, data quality checks, and target correlation audits for weather prediction datasets.

```python
import os
import sys
import pandas as pd
import numpy as np

BASE_DIR = os.path.dirname(os.getcwd()) if os.path.basename(os.getcwd()) == 'notebooks' else os.getcwd()
sys.path.insert(0, BASE_DIR)

from ml.config import WEATHER_DATASET_XLSX, WEATHER_DATASET_CSV
from ml.data_quality import audit_dataset_quality

path = WEATHER_DATASET_CSV if os.path.exists(WEATHER_DATASET_CSV) else WEATHER_DATASET_XLSX
df_rain = pd.read_excel(path) if path.endswith('.xlsx') else pd.read_csv(path)
print(f"Weather Dataset Shape: {df_rain.shape}")
df_rain.head()
```

### 1. Date Range & Seasonality Verification

```python
if 'date_of_record' in df_rain.columns:
    df_rain['date'] = pd.to_datetime(df_rain['date_of_record'], errors='coerce')
    print("Min Date:", df_rain['date'].min(), "| Max Date:", df_rain['date'].max())
    print("Records by Year:\n", df_rain['date'].dt.year.value_counts().sort_index())
```

### 2. Data Quality & Target Leakage Audit

```python
df_rain['Rain_Target'] = (df_rain['rainfall'] > 0).astype(int)
audit_res = audit_dataset_quality(df_rain, target_col='Rain_Target', name='Weather Prediction Dataset')
print("Duplicate Count:", audit_res['duplicate_rows'])
print("Missing Values:", audit_res['missing_values'])
```
